## **1. Missing Values**


**Understand the Concept**

Missing values are data values that are not available in a dataset.



**Demonstrate the Concept**

Missing values can appear as empty cells or null values.

**Real-Life Example**

An employee dataset may have missing values if an employee's age, salary, department, or email is not recorded.

**AI/ML Usage**

Missing values can affect data analysis and Machine Learning models. They should be identified before preprocessing.

## **1.1 Number of Missing Values**



**Understand the Concept**

The number of missing values shows how many data values are missing in the dataset.



**Demonstrate the Concept**

We can count all missing values to understand the overall amount of missing data.


**Real-Life Example**

An employee dataset may have missing values in Age, Salary, Department, City, or Email.

**AI/ML Usage**

Knowing the total number of missing values helps us decide whether the missing data should be removed, filled, or handled in another way.

```python
# no of missing values
df.isnull().sum()
```

```python
# total no of missing values
df.isnull().sum().sum()
```

**Interpretation**

- The output shows the total number of missing values in the dataset.
- This helps us understand the overall amount of missing data.

## **1.2. Percentage of Missing Values**



**Understand the Concept**

The percentage of missing values shows how much data is missing compared to the total data.



**Demonstrate the Concept**

It helps us understand the amount of missing data in each column as a percentage.

**Real-Life Example**

If 10 values are missing out of 1,000 records, the missing value percentage is 1%.

**AI/ML Usage**

The missing value percentage helps us decide whether to remove, fill, or keep missing data during preprocessing.

```python
# missing value percentage for each column
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)
```

**Interpretation**

- The output shows the percentage of missing values in each column.
- A higher percentage means more data is missing in that column.

## **1.3 Missing Values by Column**



**Understand the Concept**

Missing values by column show how many values are missing in each column.



**Demonstrate the Concept**

This helps us identify which columns contain missing data.

**Real-Life Example**

In an employee dataset, Age may have 5 missing values while Salary may have 3 missing values.

**AI/ML Usage**

Checking missing values by column helps us choose the correct method to handle missing data for each feature.

```python
# Missing Values by Column
missing_by_column = df.isnull().sum()
print(missing_by_column)
```

**Interpretation**

- The output shows missing values for each column.
- Columns with 0 have no missing values.
- Columns with values greater than 0 contain missing data.

## **1.4. Missing Value Patterns**



**Understand the Concept**

Missing value patterns show how missing values are distributed in a dataset.



**Demonstrate the Concept**

It helps us check whether missing values appear in the same rows or different columns.

**Real-Life Example**

An employee may have both Age and Salary missing, while another employee may only have the Email missing.

**AI/ML Usage**

Missing value patterns help us understand the relationship between missing values and choose a suitable preprocessing method.

```python
# Missing Value Patterns 
missing_rows = df[df.isnull().any(axis=1)]
missing_rows
```

**Interpretation**

- The output displays rows that contain at least one missing value.
- We can check whether multiple columns have missing values in the same row.

## **2. Duplicate Records**



**Understand the Concept**

Duplicate records are rows that appear more than once in a dataset.

**AI/ML Usage**

Duplicate records can affect data analysis and Machine Learning models. They should be identified before preprocessing.

## **2.1. Identify Duplicates**



**Understand the Concept**

Identifying duplicates means finding rows that are repeated in a dataset.

**Demonstrate the Concept**

The duplicated() function checks whether a row already appears earlier in the dataset.


**Real-Life Example**

The same employee record may be entered more than once by mistake.

**AI/ML Usage**

Duplicate records can affect data analysis and Machine Learning results. They should be identified before preprocessing.

```python
# duplicate rows
duplicate_records = df[df.duplicated()]
duplicate_records
```

**Interpretation**

- The output displays the duplicate rows.
- Duplicate records are repeated employee records.

## **2.2 Count Duplicates**



**Understand the Concept**

Counting duplicates shows how many duplicate rows are present in a dataset.

**Demonstrate the Concept**

The duplicated() function identifies duplicate rows, and sum() counts them.


**Real-Life Example**

If the same employee record appears multiple times, the repeated rows are counted as duplicates.

**AI/ML Usage**

Duplicate records can affect data analysis and Machine Learning results. Counting them helps us understand the data quality.

```python
# Count duplicate rows
duplicate_count = df.duplicated().sum()
print("No of duplicate rows:", duplicate_count)
```

**Interpretation**

- The output shows the total number of duplicate rows.

## **2.3 Completely Duplicated Rows**



**Understand the Concept**

Completely duplicated rows are rows where all column values are exactly the same.

**Demonstrate the Concept**

We can compare all columns in each row to find records that are completely repeated.

**Real-Life Example**

If the same employee record, including EmployeeID, Name, Age, Department, and Salary, appears more than once, it is a completely duplicated row.

**AI/ML Usage**

Completely duplicated rows can create repeated information and affect data analysis and Machine Learning results.

```python
# completely duplicated rows
completely_duplicated = df[df.duplicated(keep=False)]
completely_duplicated
```

**Interpretation**

- The output displays all rows that are completely duplicated.
- All column values in these rows are the same.

## **2.4 Potential Duplicate Records**



**Understand the Concept**

Potential duplicate records are records that may represent the same person but are not completely identical.

**Demonstrate the Concept**

Two records may have the same Name, Email, and other details but different EmployeeID values.


**Real-Life Example**

The same employee may be entered twice with different EmployeeID values due to a data entry mistake.

**AI/ML Usage**

Potential duplicates can create repeated information and affect data analysis and Machine Learning results.

```python
# potential duplicates based on employee details
potential_duplicates = df[df.duplicated(subset=["Name", "Email"],keep=False)]
potential_duplicates
```

**Interpretation**

- The output displays records with the same Name and Email.
- These records may represent the same employee.
- They should be reviewed to confirm whether they are actual duplicates.

## **3. Invalid Values**

## **3.1 Negative Values**




**Understand the Concept**

Negative values are numbers below zero that may not be valid for certain columns.


**Demonstrate the Concept**

Columns such as Age, Salary, and Experience should not normally contain negative values.


**Real-Life Example**

An employee cannot have an age of -5 or a salary of -10,000.

**AI/ML Usage**

Invalid negative values can affect data analysis and Machine Learning models.

```python
# negative values in Age, Salary, and Experience
negative_values = df[(pd.to_numeric(df["Age"], errors="coerce") < 0) |
                     (pd.to_numeric(df["Salary"], errors="coerce") < 0) |
                     (pd.to_numeric(df["Experience"], errors="coerce") < 0)]
negative_values
```

**Interpretation**

- The output displays records with invalid negative values.
- Negative values are present in columns where they are not appropriate.

## **3.2 Impossible Values**



**Understand the Concept**

Impossible values are values that are outside the valid or expected range.



**Demonstrate the Concept**

For example, an employee cannot normally have an age of 150, 90 years of experience, or a performance score of 10 when the valid score is from 1 to 5.

**Real-Life Example**

An employee with 150 years of age is an impossible value and indicates a data quality issue.

**AI/ML Usage**

Impossible values can affect data analysis and Machine Learning results. 

```python
# Convert columns to numeric values
age = pd.to_numeric(df["Age"], errors="coerce")
experience = pd.to_numeric(df["Experience"], errors="coerce")
performance = pd.to_numeric(df["PerformanceScore"], errors="coerce")

# impossible values
impossible_values = df[(age > 100)|(experience > 50)|(performance < 1)|(performance > 5)]
impossible_values
```

**Interpretation**

- The output displays records with impossible values.
- These values are outside the expected range.

## **3.3 Incorrect Categories**



**Understand the Concept**

Incorrect categories are values that do not belong to the expected categories of a column.



**Demonstrate the Concept**

For example, the Gender column may be expected to contain only Male or Female.



**Real-Life Example**

If Gender contains values such as M, Other, or Not Available, these can be treated as incorrect or unexpected categories based on the dataset rules.

**AI/ML Usage**

Incorrect categories can create errors during data analysis and encoding. 

```python
# incorrect values in Gender col
valid_gender = ["Male", "Female"]

incorrect_categories = df[~df["Gender"].isin(valid_gender)]
incorrect_categories
```

**Interpretation**

- The output displays records with incorrect Gender categories.
- These values do not match the expected categories.

## **3.4 Invalid Dates**



**Understand the Concept**

Invalid dates are dates that do not follow a valid date format or do not exist.



**Demonstrate the Concept**

For example, 2025-13-40 and 2024-02-30 are invalid dates.

**Real-Life Example**

An employee joining date must be a valid calendar date. A date with an invalid month or day indicates a data quality issue.

**AI/ML Usage**

Invalid dates can cause errors during data processing and feature creation. 

```python
# Convert JoiningDate to datetime
joining_date = pd.to_datetime( df["JoiningDate"],errors="coerce")

# invalid dates
invalid_dates = df[joining_date.isnull()]
invalid_dates
```

## **3.5 Incorrect Data Types**



**Understand the Concept**

Incorrect data types occur when a column contains a value with an unsuitable data type.



**Demonstrate the Concept**

For example, Age should contain numbers, but a value such as Thirty is text.

**Real-Life Example**

Salary and Experience should contain numerical values. Text values in these columns can cause data quality issues.

**AI/ML Usage**

Incorrect data types can cause errors during data analysis and Machine Learning.

```python
# Convert expected numerical columns
age = pd.to_numeric(df["Age"], errors="coerce")
salary = pd.to_numeric(df["Salary"], errors="coerce")
experience = pd.to_numeric(df["Experience"], errors="coerce")

# rows with incorrect data types
incorrect_data_types = df[age.isnull() | salary.isnull() | experience.isnull()]
incorrect_data_types
```

**Interpretation**

- The output displays records with possible incorrect data types.
- Text values are present in columns expected to contain numbers.

## **3.6 Unexpected Values**



**Understand the Concept**

Unexpected values are values that do not match the expected data or format.



**Demonstrate the Concept**

For example, the Status column is expected to contain Active, Inactive, or On Leave. Other values are unexpected.

**Real-Life Example**

If an employee status contains Working instead of the expected categories, it is an unexpected value.

**AI/ML Usage**

Unexpected values can affect data analysis and Machine Learning results.

```python
# expected values
valid_status = ["Active", "Inactive", "On Leave"]

# Find unexpected values
unexpected_values = df[~df["Status"].isin(valid_status)]
unexpected_values
```

**Interpretation**

- The output displays records with unexpected Status values.
- These values do not match the expected categories.

## **3.7 Empty Strings**



**Understand the Concept**

Empty strings are text values that contain no information.

**Demonstrate the Concept**

An empty string may appear as a blank value in a text column.


**Real-Life Example**

An employee Name or City field may be left blank during data entry.

**AI/ML Usage**

Empty strings can affect data analysis and Machine Learning.

```python
# rows containing empty strings
empty_strings = df[df.select_dtypes(include="object").eq("").any(axis=1)]
empty_strings
```

**Interpretation**

- The output displays rows containing empty strings.
- Empty strings represent missing or unavailable text data.

## **3.8 Null-Like Values**



**Understand the Concept**

Null-like values are text values that represent missing or unavailable data.



**Demonstrate the Concept**

Common null-like values include NA, N/A, NULL, None, and Unknown.


**Real-Life Example**

An employee Department may contain N/A instead of an actual department name.

**AI/ML Usage**

Null-like values should be identified and converted to proper missing values before data analysis & Machine Learning.

```python
# common null-like values
null_like_values = ["NA", "N/A", "NULL", "None", "Unknown"]

# rows containing null-like values
null_like_rows = df[ df.astype(str).isin(null_like_values).any(axis=1)]
null_like_rows
```

**Interpretation**

- The output displays rows containing null-like values.
- These values represent missing or unavailable information.
