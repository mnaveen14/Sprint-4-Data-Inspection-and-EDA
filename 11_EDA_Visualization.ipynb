# **Notebook 11 – EDA Visualization**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("employee_salary_dataset.csv")
```

## **1. Histogram**

A histogram shows the distribution of numerical data using grouped intervals.

**Real-Life Example**

It can show how employee salaries are distributed across different salary ranges.

**AI/ML Usage**

It helps identify distribution patterns, skewness, spread, and possible outliers before model training.

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="MonthlySalary", bins=30, kde=True)
plt.title("Distribution of Monthly Salary")
plt.xlabel("Monthly Salary")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

**Interpretation**

- The histogram shows how MonthlySalary values are distributed.
- It helps identify the concentration and spread of employee salaries.
- The KDE curve shows the overall shape of the distribution.

**Insight Obtained**

The visualization helps identify the common salary range and whether the salary distribution contains skewness or unusually high or low values.

## **2. Box Plot**


A box plot shows the spread, median, quartiles, and potential outliers of numerical data.

**Real-Life Example**

It can show whether some employees have unusually high or low monthly salaries.

**AI/ML Usage**

It helps detect outliers and understand data spread before model training.

```python
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x="MonthlySalary")
plt.title("Box Plot of Monthly Salary")
plt.xlabel("Monthly Salary")
plt.tight_layout()
plt.show()
```

**Interpretation**

- The box shows the middle 50% of salary values.
- The line inside the box represents the median.
- Points outside the whiskers indicate potential outliers.

**Insight Obtained**

The visualization helps identify unusual salary values that may require further investigation.

## **3. Bar Chart**


A bar chart compares values across different categories.

**Real-Life Example**

It can compare the average monthly salary across different departments.

**AI/ML Usage**

It helps identify differences and patterns between categorical groups.

```python
# Calculate average monthly salary by department

average_salary = (df.groupby("Department")["MonthlySalary"].mean().sort_values(ascending=False))
plt.figure(figsize=(8, 5))
average_salary.plot(kind="bar")
plt.title("Average Monthly Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Monthly Salary")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Interpretation**

- Each bar represents the average monthly salary of a department.
- Taller bars indicate departments with higher average salaries.


## **4. Count Plot**


A count plot shows the number of observations in each category.

**Real-Life Example**

It can show how many employees belong to each department.

**AI/ML Usage**

It helps identify category distribution, dominant categories, and possible category imbalance.

```python
plt.figure(figsize=(8, 5))
sns.countplot( data=df, x="Department")
plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Interpretation**

- Each bar represents the number of employees in a department.
- Taller bars indicate departments with more employees.

**Insight Obtained**

The chart helps understand how employees are distributed across different departments.

## **5. Scatter Plot**


A scatter plot shows the relationship between two numerical variables.

**Real-Life Example**

It can show how employee experience is related to monthly salary.

**AI/ML Usage**

It helps identify relationships, trends, correlations, and possible outliers between numerical features.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="ExperienceYears", y="MonthlySalary")
plt.title("Experience Years vs Monthly Salary")
plt.xlabel("Experience Years")
plt.ylabel("Monthly Salary")
plt.tight_layout()
plt.show()
```

**Interpretation**

- Each point represents an employee.
- The plot shows how MonthlySalary changes with ExperienceYears.
- An upward trend indicates that salary generally increases with experience.

**Insight Obtained**

The visualization helps determine whether employee experience has a meaningful relationship with monthly salary.

## **6. Line Plot**


A line plot shows trends or changes in values across an ordered variable.

**Real-Life Example**

It can show how the average monthly salary changes across different years of employee experience.

**AI/ML Usage**

It helps identify trends and patterns between ordered numerical variables.

```python
# average salary by experience
average_salary = (df.groupby("ExperienceYears")["MonthlySalary"].mean())
plt.figure(figsize=(8, 5))
plt.plot(average_salary.index, average_salary.values, marker="o")
plt.title("Average Monthly Salary by Experience")
plt.xlabel("Experience Years")
plt.ylabel("Average Monthly Salary")
plt.tight_layout()
plt.show()
```

**Interpretation**

- The line shows how average salary changes with experience.
- An upward trend indicates that employees with more experience generally earn higher salaries.

**Insight Obtained**

Employee experience appears to have a positive relationship with average monthly salary.

## **7. Violin Plot**

A violin plot shows the distribution and density of numerical data across different categories.

**Real-Life Example**

It can compare the distribution of monthly salaries across different departments.

**AI/ML Usage**

It helps compare data distributions, spread, and patterns between categories.

```python
plt.figure(figsize=(9, 5))
sns.violinplot( data=df, x="Department",    y="MonthlySalary")
plt.title("Monthly Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Monthly Salary")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Interpretation**

- The width of each violin shows where salary values are more concentrated.
- Wider sections represent a higher density of employees in that salary range.
- The plot compares salary distributions across departments.

**Insight Obtained**

The visualization helps identify departments with different salary patterns, spreads, and concentrations.

## **8. Pair Plot**


A pair plot shows relationships and distributions between multiple numerical variables.

**Real-Life Example**

It can compare Age, ExperienceYears, WorkingHours, PerformanceScore, and MonthlySalary.

**AI/ML Usage**

It helps identify correlations, trends, clusters, and potential relationships between multiple numerical features.

```python
#  numerical columns
numerical_features = ["Age","ExperienceYears","WorkingHours","OvertimeHours","PerformanceScore","MonthlySalary"]

sns.pairplot(df[numerical_features])
plt.show()
```

**Interpretation**

- Diagonal plots show the distribution of individual numerical variables.
- Other plots show relationships between pairs of variables.
- Patterns may indicate positive, negative, or weak relationships.

**Insight Obtained**

The pair plot provides an overall view of relationships between important numerical features and helps identify features that may be related to MonthlySalary.

## **9. Heatmap**


A heatmap visually represents the correlation between multiple numerical variables using color intensity.

**Real-Life Example**

It can show how Age, ExperienceYears, PerformanceScore, and MonthlySalary are related to each other.

**AI/ML Usage**

It helps identify strong correlations, weak relationships, and potential multicollinearity before model training.

```python
# correlation matrix
correlation_matrix = df[numerical_features].corr()

# heatmap
plt.figure(figsize=(8, 6))
sns.heatmap( correlation_matrix, annot=True, fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

**Interpretation**

- Values close to 1 indicate a strong positive relationship.
- Values close to -1 indicate a strong negative relationship.
- Values close to 0 indicate a weak relationship.

**Insight Obtained**

The heatmap helps identify which numerical features are strongly related to `MonthlySalary` and whether any features have potential multicollinearity.

## **10. Distribution Plot**



A distribution plot shows the distribution and overall pattern of a numerical variable.

**Real-Life Example**

It can show how employee monthly salaries are distributed across different salary ranges.

**AI/ML Usage**

It helps identify the shape, center, spread, skewness, and possible outliers of numerical data.

```python
plt.figure(figsize=(8, 5))
sns.histplot( data=df, x="MonthlySalary", kde=True, bins=30)
plt.title("Distribution Plot of Monthly Salary")
plt.xlabel("Monthly Salary")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

**Interpretation**

- The histogram shows the frequency of salary values.
- The KDE curve shows the overall distribution pattern.
- The plot helps identify whether the distribution is symmetric or skewed.

**Insight Obtained**

The visualization provides a clear understanding of the overall distribution of MonthlySalary and helps identify unusual patterns before Machine Learning.
