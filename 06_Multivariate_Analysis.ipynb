# **Notebook 6 – Multivariate Analysis**

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("Employee_Salary_Performance_Dataset.csv")
df.info()
```

## **1. Correlation Matrix**



**Understand the Concept**

A correlation matrix shows the relationship between multiple numerical variables.

Correlation values usually range from **-1 to +1**:

- **+1** - Perfect positive correlation
- **0** - No linear relationship
- **-1** - Perfect negative correlation



**Demonstrate the Concept**

We can analyze the relationships between Age, Experience, Salary, TrainingHours, PerformanceScore

**Real-Life Example**

- Experience and Salary may have a positive correlation because employees with more experience may earn higher salaries.
- At the same time, some variables may have weak or negative correlations.

**AI/ML Usage**

Correlation analysis helps us:

- Identify strongly related features.
- Identify weak relationships.
- Detect negative correlations.
- Identify potential multicollinearity between input features.
- Support feature selection before building Machine Learning models.

```python
# numerical columns
numerical_columns = df.select_dtypes(include="number").columns

# correlation matrix
correlation_matrix = df[numerical_columns].corr()
print(correlation_matrix.round(2))
```

**Interpretation**

- Correlation values close to **+1** indicate a strong positive relationship.
- Correlation values close to **-1** indicate a strong negative relationship.
- Correlation values close to **0** indicate a weak or no linear relationship.
- If two input features have a very high correlation, it may indicate potential multicollinearity.

## **2. Heatmap**



**Understand the Concept**

A heatmap is a visual representation of the correlation matrix.



**Demonstrate the Concept**

It uses different color intensities to show the strength and direction of relationships between multiple numerical variables.

**Real-Life Example**

Using the employee dataset, we can visualize the relationships between Age, Experience, Salary, TrainingHours, PerformanceScore

This makes it easier to identify strong, weak, and negative correlations.

**AI/ML Usage**

A correlation heatmap helps us:

- Identify strongly correlated features.
- Identify weak relationships.
- Detect negative correlations.
- Identify potential multicollinearity.
- Support feature selection before building Machine Learning models.

```python
# heatmap for the correlation matrix
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm" )
plt.title("Correlation Heatmap")
plt.show()
```

**Interpretation**

- Features with very high correlations may indicate potential multicollinearity.

## **3. Pair Plot**



**Understand the Concept**

A pair plot visualizes the relationships between multiple numerical variables at the same time.



**Demonstrate the Concept**

It creates:

- Scatter plots between pairs of numerical variables.
- Distribution plots for individual variables.

This helps us analyze multiple feature relationships in one visualization.

**Real-Life Example**

 we can observe whether Experience and Salary have a relationship and compare it with other feature relationships.

**AI/ML Usage**

Pair plots help us:

- Identify strong and weak relationships.
- Detect positive and negative relationships.
- Identify possible patterns between features.
- Detect potential multicollinearity.
- Understand numerical features before building a Machine Learning model.

```python
# numerical columns
numerical_data = df.select_dtypes(include="number")

# pair plot
sns.pairplot(numerical_data)
plt.show()
```

**Interpretation**

- Scatter plots show the relationship between every pair of numerical variables.
- The diagonal plots show the distribution of individual variables.
- Strong relationships may show a clear upward or downward pattern.
- Weak relationships may appear as randomly scattered points.

## **4. Multiple Feature Relationships**



**Understand the Concept**

Multiple Feature Relationships analyze how more than two variables are related to each other at the same time.



**Demonstrate the Concept**

Instead of analyzing only two variables, we can examine multiple features together to identify patterns and relationships.

**Real-Life Example**

Employees with more Experience and higher PerformanceScore may tend to have higher Salary.

**AI/ML Usage**

Multiple feature relationship analysis helps us:

- Identify important features.
- Understand relationships between multiple variables.
- Detect possible feature interactions.
- Identify redundant features.
- Support feature selection and feature engineering before building Machine Learning models.

```python
# Analyze the relationship between multiple numerical features
sns.pairplot(df[["Age", "Experience", "Salary", "TrainingHours", "PerformanceScore"]])
plt.show()
```

**Interpretation**

- The plots help us observe relationships between multiple numerical features.
- We can identify which variables may have stronger or weaker relationships with Salary.
- Clear upward patterns may indicate positive relationships.
- Clear downward patterns may indicate negative relationships.
- Random scatter may indicate weak relationships.

## **5. Feature Interactions**



**Understand the Concept**

Feature interaction occurs when the relationship between one feature and another feature changes depending on a third feature.



**Demonstrate the Concept**

Instead of analyzing Salary only with Experience, we can also consider another feature such as Department.

For example:

- Experience vs Salary
- Department as an additional variable

This helps us understand whether the relationship between Experience and Salary is different across departments.

**Real-Life Example**

Employees with more experience may generally have higher salaries, but the salary growth may vary between different departments.

**AI/ML Usage**

Feature interactions help us:

- Identify combined effects of multiple features.
- Discover hidden patterns in the data.
- Create new interaction features.
- Improve model performance.
- Help tree-based models capture complex relationships.

```python
# Analyze the interaction between Experience, Salary, Department
plt.figure(figsize=(9, 5))
sns.scatterplot( data=df, x="Experience", y="Salary", hue="Department",  alpha=0.7)
plt.title("Experience vs Salary by Department")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()
```

**Interpretation**

- The scatter plot shows the relationship between Experience and Salary across different departments.
- Different colors represent different departments.
- We can observe whether the relationship between Experience and Salary changes by Department.
- Different patterns across departments may indicate a feature interaction.

## **6. Numerical Feature Relationships**



**Understand the Concept**

Numerical Feature Relationships analyze how multiple numerical variables are related to each other.



**Demonstrate the Concept**

We can examine the relationships between numerical features

This helps us identify strong, weak, positive, and negative relationships between features.

**Real-Life Example**

In the employee dataset:

- Experience may have a positive relationship with Salary.
- Age and Experience may also be related.
- TrainingHours may have a weak or strong relationship with PerformanceScore.

By analyzing these relationships together, we can better understand the dataset.

**AI/ML Usage**

Numerical feature relationship analysis helps us:

- Identify important features.
- Detect strong and weak correlations.
- Identify negative relationships.
- Detect potential multicollinearity.
- Support feature selection and feature engineering.

```python
# correlation with Salary
salary_relationship = numerical_data.corr()["Salary"].sort_values( ascending=False)
print(salary_relationship.round(2))
```

**Interpretation**

- The output shows how each numerical feature is related to Salary.
- Positive values indicate that both variables tend to increase together.
- Negative values indicate that one variable may decrease as the other increases.
- Values close to 0 indicate a weak relationship.
- Features with strong relationships with Salary may be useful for prediction models.
- Very high correlations between input features may indicate potential multicollinearity.
