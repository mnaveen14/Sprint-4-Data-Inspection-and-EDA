# **Notebook 12 – Business Insights**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("employee_salary_dataset.csv")
```

## **1. Important Patterns**



```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df,x="ExperienceYears",y="MonthlySalary")
plt.title("Experience Years vs Monthly Salary")
plt.xlabel("Experience Years")
plt.ylabel("Monthly Salary")
plt.tight_layout()
plt.show()
```

**Observation**

MonthlySalary generally increases with ExperienceYears and PerformanceScore. Salary levels also vary across different departments and education levels.

**Business Insight**

Employee experience, performance, department, and education appear to be important factors influencing salary. These patterns can help the organization understand its overall compensation structure.

**Potential Action**

The company can review salary patterns across departments and employee groups to ensure consistent and fair compensation planning.

## **2. Trends**

```python
# Calculate average salary by experience

average_salary = (df.groupby("ExperienceYears")["MonthlySalary"].mean().reset_index())
plt.figure(figsize=(8, 5))
sns.lineplot(data=average_salary, x="ExperienceYears", y="MonthlySalary", marker="o")
plt.title("Average Monthly Salary by Experience")
plt.xlabel("Experience Years")
plt.ylabel("Average Monthly Salary")
plt.tight_layout()
plt.show()
```

**Observation**

Average MonthlySalary generally increases as ExperienceYears increases.

**Business Insight**

Employee compensation tends to grow with experience, indicating that experienced employees represent a higher salary investment for the organization.

**Potential Action**

The organization can use this trend for salary forecasting, workforce planning, and compensation budgeting.

## **3. Relationships**

**Visualization**

A scatter plot can show the relationship between PerformanceScore and MonthlySalary.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df,x="PerformanceScore", y="MonthlySalary")
plt.title("Performance Score vs Monthly Salary")
plt.xlabel("Performance Score")
plt.ylabel("Monthly Salary")
plt.tight_layout()
plt.show()
```

**Observation**

MonthlySalary generally increases as PerformanceScore increases.

**Business Insight**

Employee performance appears to have a positive relationship with salary, suggesting that higher-performing employees receive higher compensation.

**Potential Action**

The organization can review performance-based compensation policies to ensure that salary growth is aligned with employee performance.

## **4. Anomalies**

**Visualization**

A box plot can identify unusually high or low values in MonthlySalary.

```python
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x="MonthlySalary")
plt.title("Monthly Salary Anomalies")
plt.xlabel("Monthly Salary")
plt.tight_layout()
plt.show()
```

**Observation**

Some employees may have unusually high or low salary values compared to the overall salary distribution.

**Business Insight**

Unusual salary values may represent senior roles, special compensation packages, or possible data inconsistencies.

**Potential Action**

The organization should review these records to determine whether the values are valid or require correction.

## **5. Potential Business Problems**

**Visualization**

A bar chart can compare the average MonthlySalary across different departments.

```python
# average salary by department

average_salary = (df.groupby("Department")["MonthlySalary"].mean().sort_values(ascending=False))
plt.figure(figsize=(8, 5))
average_salary.plot(kind="bar")
plt.title("Average Monthly Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Monthly Salary")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Observation**

Average MonthlySalary varies across different departments.

**Business Insight**

Large salary differences between departments may create compensation inequality or indicate differences in skill requirements and market demand.

**Potential Action**

The organization can review department-wise compensation structures to identify unjustified salary gaps and improve compensation planning.

## **6. Interesting Segments**

**Visualization**

A violin plot can compare the distribution of MonthlySalary across different departments.

```python
plt.figure(figsize=(9, 5))
sns.violinplot(data=df, x="Department", y="MonthlySalary")
plt.title("Monthly Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Monthly Salary")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Observation**

Different departments show different salary distributions and ranges.

**Business Insight**

Some departments may represent high-salary segments, while others may have lower or more concentrated salary ranges.

**Potential Action**

The organization can analyze these employee segments separately for compensation planning, budgeting, and talent retention.

## **7. Potential Opportunities**

**Visualization**

A bar chart can compare the average PerformanceScore across different departments.

```python
# average performance score by department
average_performance = (df.groupby("Department")["PerformanceScore"].mean().sort_values(ascending=False))
plt.figure(figsize=(8, 5))
average_performance.plot(kind="bar")
plt.title("Average Performance Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Performance Score")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Observation**

Average PerformanceScore varies across different departments.

**Business Insight**

Departments with higher performance may have effective practices that could be studied and applied to other departments.

**Potential Action**

The organization can identify successful department practices and use them to improve employee performance across other teams.
