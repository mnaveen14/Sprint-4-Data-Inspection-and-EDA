# **Notebook 5 – Bivariate Analysis**

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("Employee_Salary_Performance_Dataset.csv")
df.info()
```

## **Bivariate Analysis**


- Bivariate Analysis studies the relationship between two variables.
- It helps us understand how one variable changes or is related to another variable.



### **1. Numerical vs Numerical**



**Understand the Concept**

Numerical vs Numerical analysis studies the relationship between two numerical variables.

**Demonstrate the Concept**

We can check whether changes in one numerical variable are related to changes in another variable.

**Real-Life Example**

we analyze Experience and Salary to understand whether employees with more experience tend to have higher salaries.

**AI/ML Usage**

This analysis helps identify relationships between numerical features and detect highly correlated variables before building Machine Learning models.

```python
# Scatter plot - Experience vs Salary
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Experience",  y="Salary",  alpha=0.6)
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()
```

- The scatter plot shows the relationship between Experience and Salary.
- Employees with more experience tend to have higher salaries.

```python
# Scatter plot - Age vs Salary
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Age", y="Salary", alpha=0.6)
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

The scatter plot shows the relationship between employee Age and Salary.

```python
# correlation between numerical variables
correlation = df[["Age", "Experience", "Salary", "PerformanceScore", "TrainingHours"]].corr()
print(correlation.round(2))
```

- The correlation matrix helps identify the strength of relationships between numerical variables.
- Highly correlated features may need attention during Machine Learning model building.

### **2. Numerical vs Categorical**



**Understand the Concept**

Numerical vs Categorical analysis compares a numerical variable across different categories.

**Demonstrate the Concept**

It helps us understand how numerical values differ between groups.

**Real-Life Example**

Analyze Department vs Salary and Status vs Salary to compare employee salaries across different categories.

**AI/ML Usage**

This analysis helps identify important category-based patterns and differences that can be useful for feature engineering and Machine Learning models.

```python
# Box plot - Department vs Salary
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="Department", y="Salary")
plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.show()
```

**Interpretation**

- The box plot compares salary distributions across departments.
- It helps identify differences in median salary, spread, and possible outliers.

```python
# Violin plot - Status vs Salary
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x="Status", y="Salary")
plt.title("Salary Distribution by Employee Status")
plt.xlabel("Status")
plt.ylabel("Salary")
plt.show()
```

**Interpretation**
- The violin plot shows the salary distribution and density for each employee status.
- These comparisons help us understand how Salary varies across different categories.

### **3. Categorical vs Categorical**



**Understand the Concept**

Categorical vs Categorical analysis studies the relationship between two categorical variables.

**Demonstrate the Concept**

It helps us compare the frequency and distribution of one category across another category.

**Real-Life Example**

Analyze Department vs Status and Gender vs Status to understand how employee status is distributed across different categories.

**AI/ML Usage**

This analysis helps identify relationships between categorical features and can be useful for feature selection and classification models.

```python
# Grouped bar chart - Department vs Status
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x="Department", hue="Status")
plt.title("Employee Status by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

**Interpretation**

- The grouped bar chart compares employee status across departments.
- It helps identify differences in the number of Active, Inactive, and On Leave employees.


```python
# Cross tabulation - Gender vs Status
cross_tab = pd.crosstab(df["Gender"], df["Status"])
print(cross_tab)
```

**Interpretation**
- The cross tabulation shows the frequency of each Status for Male and Female employees.
- These techniques help us understand relationships between categorical variables.
