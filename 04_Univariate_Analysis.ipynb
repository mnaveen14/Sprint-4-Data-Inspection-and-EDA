## **Univariate Analysis**

**Understand the Concept**

Univariate Analysis means analyzing one variable at a time.

**Demonstrate the Concept**

We study a single column to understand its values, distribution, and characteristics.


**AI/ML Usage**

Univariate analysis helps us understand individual features before building a Machine Learning model. It can also help identify data patterns, unusual values, and outliers.

### **Numerical Variables**

#### **1. Mean**


**Understand the Concept**

Mean is the average value of a numerical variable.

**Demonstrate the Concept**

It is calculated by adding all values and dividing by the total number of values.


**Real-Life Example**

We can find the average Salary of all employees in the dataset.

**AI/ML Usage**

Mean helps us understand the central value of numerical data and is useful during preprocessing.

```python
# Mean - salary col
mean_salary = df["Salary"].mean()
print(mean_salary)
```

**Interpretation**

- The mean shows the average employee salary.

#### **2. Median**

**Understand the Concept**

Median is the middle value of a numerical variable after arranging the values in order.

**Demonstrate the Concept**

If the number of values is even, the median is the average of the two middle values.

**Real-Life Example**

We can find the median Salary to understand the middle salary value of employees.

**AI/ML Usage**

Median is useful for understanding the central value of data and is less affected by extreme values or outliers.

```python
# Median - salary col
median_salary = df["Salary"].median()
print(median_salary)
```

**Interpretation**

- The median shows the middle employee salary.
- Half of the salary values are below it, and half are above it

#### **3. Mode**



**Understand the Concept**

Mode is the value that appears most frequently in a variable.

**Demonstrate the Concept**

A variable can have one mode, multiple modes, or no repeated values.

**Real-Life Example**

We can find the most common PerformanceScore among employees.

**AI/ML Usage**

Mode is useful for understanding the most common value and can also be used to fill missing values in categorical or numerical data.

```python
# mode - PerformanceScore
mode_score = df["PerformanceScore"].mode()[0]
print(mode_score)
```

**Interpretation**

- The output shows the most frequently occurring PerformanceScore.
- More than one mode may exist if multiple values have the same highest frequency.

#### **4. Minimum**



**Understand the Concept**

Minimum is the smallest value in a numerical variable.

**Demonstrate the Concept**

It helps us find the lowest value present in a column.

**Real-Life Example**

We can find the minimum Salary to identify the lowest salary in the dataset.

**AI/ML Usage**

Minimum values help us understand the data range and identify unusually low values.

```python
# minimum salary
minimum_salary = df["Salary"].min()
print(minimum_salary)
```

**Interpretation**

- The output shows the lowest salary in the dataset.

#### **5. Maximum**



**Understand the Concept**

Maximum is the largest value in a numerical variable.

**Demonstrate the Concept**

It helps us find the highest value present in a column.

**Real-Life Example**

We can find the maximum Salary to identify the highest salary in the dataset.

**AI/ML Usage**

Maximum values help us understand the data range and identify unusually high values.

```python
# maximum salary
maximum_salary = df["Salary"].max()
print("Maximum Salary:", maximum_salary)
```

**Interpretation**

- The output shows the highest salary in the dataset.

#### **6. Range**



**Understand the Concept**

Range shows the difference between the maximum and minimum values.

**Demonstrate the Concept**

It helps us understand the overall spread of a numerical variable.

**Real-Life Example**

We can calculate the salary range to understand the difference between the highest and lowest employee salary.

**AI/ML Usage**

Range helps us understand the spread of data and identify large differences between values.

```python
# range of Salary
salary_range = df["Salary"].max() - df["Salary"].min()
print("Salary Range:", salary_range)
```

**Interpretation**

- The output shows the difference between the highest and lowest salary.
- A larger range indicates a wider spread of salary values.

#### **7. Variance**



**Understand the Concept**

Variance measures how much the values are spread out from the mean.

**Demonstrate the Concept**

A higher variance means the values are more spread out. A lower variance means the values are closer to the mean.

**Real-Life Example**

We can calculate the variance of Salary to understand how much employee salaries differ from the average salary.

**AI/ML Usage**

Variance helps us understand data spread and is useful during data analysis and feature scaling.

```python
# variance of Salary
salary_variance = df["Salary"].var()
print("Salary Variance:", salary_variance)
```

**Interpretation**

- The output shows how much salary values vary from the mean.
- A higher variance indicates a wider spread of salaries.
- A lower variance indicates that salaries are closer to the average.

#### **8. Standard Deviation**



**Understand the Concept**

Standard deviation measures how far the values are spread from the mean.

**Demonstrate the Concept**

A higher standard deviation means the values are more spread out. A lower standard deviation means the values are closer to the mean.

**Real-Life Example**

We can calculate the standard deviation of Salary to understand how much employee salaries differ from the average salary.

**AI/ML Usage**

Standard deviation helps us understand data spread and is commonly used in feature scaling and data preprocessing.

```python
# Standard deviation of Salary
salary_std = df["Salary"].std()
print("Standard Deviation:", salary_std)
```

**Interpretation**

- The output shows how much salary values are spread from the mean.
- A higher value indicates a wider spread of salaries.

#### **9. Quartiles**



**Understand the Concept**

Quartiles divide numerical data into four equal parts.

**Demonstrate the Concept**

The main quartiles are:

- First Quartile (Q1) – 25% 
- Second Quartile (Q2) – 50% 
- Third Quartile (Q3) – 75% 

**Real-Life Example**

We can use quartiles to understand how employee salaries are distributed.

**AI/ML Usage**

Quartiles help us understand data distribution and are useful for identifying possible outliers using the IQR method.

```python
# quartiles for Salary
q1 = df["Salary"].quantile(0.25)
q2 = df["Salary"].quantile(0.50)
q3 = df["Salary"].quantile(0.75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

**Interpretation**

- Q1 shows the value below which 25% of salaries fall.
- Q2 represents the median salary.
- Q3 shows the value below which 75% of salaries fall.
- Quartiles help us understand the distribution of salary values.

#### **10. Percentiles**



**Understand the Concept**

Percentiles show the position of a value compared to the rest of the data.

**Demonstrate the Concept**

For example, the 25th percentile means that 25% of the values are below that value.

**Real-Life Example**

We can use percentiles to understand how employee salaries are distributed.

**AI/ML Usage**

Percentiles help us understand data distribution and identify unusually low or high values.

```python
# percentiles for Salary
percentiles = df["Salary"].quantile([0.10, 0.25, 0.50, 0.75, 0.90])
print(percentiles)
```

**Interpretation**

- The output shows Salary values at different percentile levels.

#### **11. Skewness**



**Understand the Concept**

Skewness shows whether the data distribution is symmetric or tilted to one side.

**Demonstrate the Concept**

- Skewness close to 0 – Data is approximately symmetric.
- Positive skewness – The distribution has a longer right tail.
- Negative skewness – The distribution has a longer left tail.

**Real-Life Example**

Employee salaries may have positive skewness if most employees have moderate salaries and only a few employees have very high salaries.

**AI/ML Usage**

Skewness helps us understand data distribution and may help us decide whether data transformation is needed before building a Machine Learning model.

```python
# skewness of Salary
salary_skewness = df["Salary"].skew()
print("Skewness:", salary_skewness)
```

**Interpretation**

- A value close to 0 indicates a more symmetric distribution.
- A positive value indicates the data is skewed to the right.
- A negative value indicates the data is skewed to the left.

#### **12. Kurtosis**



**Understand the Concept**

Kurtosis shows the shape of a data distribution, especially how heavy or light its tails are.

**Demonstrate the Concept**

A higher kurtosis value may indicate more extreme values or outliers in the data.

**Real-Life Example**

Employee salaries may have high kurtosis if most salaries are similar but a few employees have very high or very low salaries.

**AI/ML Usage**

Kurtosis helps us understand the distribution of data and identify the presence of extreme values.

```python
# kurtosis of Salary
salary_kurtosis = df["Salary"].kurt()
print("Kurtosis:", salary_kurtosis)
```

**Interpretation**

- A higher kurtosis may indicate more extreme values.
- A lower kurtosis indicates fewer extreme values.

### **Visualizations:**

#### **13. Histogram**



**Understand the Concept**

A histogram shows how numerical values are distributed.

**Demonstrate the Concept**

It groups numerical values into ranges and shows how many values fall into each range.

**Real-Life Example**

We can use a histogram to see how employee salaries are distributed across different salary ranges.

**AI/ML Usage**

Histograms help us understand data distribution, skewness, and possible unusual values before building a Machine Learning model.

```python
# Histogram
plt.figure(figsize=(8, 5))
plt.hist(df["Salary"], bins=20, edgecolor="black")
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
```

**Interpretation**

- The histogram shows the distribution of employee salaries.

#### **14. KDE / Density Plot**



**Understand the Concept**

A KDE plot shows the smooth distribution of numerical data.

**Demonstrate the Concept**

It estimates the density of values and shows where the data is more concentrated.

**Real-Life Example**

We can use a KDE plot to see the overall distribution of employee salaries.

**AI/ML Usage**

KDE plots help us understand data distribution, skewness, and the concentration of values before building a Machine Learning model.

```python
# KDE / Density Plot
df["Salary"].plot(kind="kde",  figsize=(8, 5))
plt.title("Salary Density Distribution")
plt.xlabel("Salary")
plt.ylabel("Density")
plt.show()
```

**Interpretation**

- The KDE plot shows the smooth distribution of employee salaries.
- Peaks represent areas where salary values are more concentrated.

#### **15. Box Plot**



**Understand the Concept**

A box plot shows the distribution, spread, and possible outliers of numerical data.

**Demonstrate the Concept**

It displays the minimum, first quartile, median, third quartile, maximum, and possible outliers.

**Real-Life Example**

We can use a box plot to understand the spread of employee salaries and identify unusually high or low salaries.

**AI/ML Usage**

Box plots help identify outliers that may affect Machine Learning models.

```python
# Box Plot
plt.figure(figsize=(8, 5))
plt.boxplot(df["Salary"])
plt.title("Salary Distribution")
plt.ylabel("Salary")
plt.show()
```

**Interpretation**

- The box plot shows the spread of employee salaries.
- The line inside the box represents the median.
- Points outside the main range may represent possible outliers.

### **Categorical Variables**

#### **16. Frequency**



**Understand the Concept**

Frequency shows how many times each value appears in a variable.

**Demonstrate the Concept**

It helps us understand how data is distributed across different categories.

**Real-Life Example**

We can find how many employees belong to each Department.

**AI/ML Usage**

Frequency analysis helps us understand categorical data and identify categories with very high or low numbers of records.

```python
# frequency of each Department
department_frequency = df["Department"].value_counts()
print(department_frequency)
```

**Interpretation**

- Some departments may have more employees than others.

#### **17. Unique Values**



**Understand the Concept**

Unique values are the different values present in a variable.

**Demonstrate the Concept**

It helps us find how many different categories or values are available in a column.

**Real-Life Example**

The Department column may contain values such as IT, HR, Finance, Sales, Marketing, and Operations.

**AI/ML Usage**

Unique values help us understand the type of a feature and decide how it should be processed before building a Machine Learning model.

```python
# unique values in Department
unique_departments = df["Department"].unique()
print(unique_departments)

# Count unique values
print("Number of unique values:", df["Department"].nunique())
```

**Interpretation**

- nunique() shows the total number of unique departments.

#### **18. Value Counts**



**Understand the Concept**

Value counts show how many times each unique value appears in a variable.

**Demonstrate the Concept**

It counts the frequency of every category in a column.

**Real-Life Example**

We can use value counts to find the number of employees in each Department.

**AI/ML Usage**

Value counts help us understand category distribution and identify categories with very high or low frequencies.

#### **19. Percentage Distribution**



**Understand the Concept**

Percentage distribution shows the percentage of each category in a variable.

**Demonstrate the Concept**

It helps us understand the proportion of each category compared to the total number of records.

**Real-Life Example**

We can find the percentage of employees in each Department.

**AI/ML Usage**

Percentage distribution helps us understand category balance and identify categories with very high or low representation.

```python
# percentage distribution of Department
department_percentage = (df["Department"].value_counts(normalize=True) * 100)
print(department_percentage)
```

**Interpretation**

- Higher percentages indicate departments with more employee records.

### **Visualizations:**

#### **20. Bar Chart**



**Understand the Concept**

A bar chart shows the frequency of different categories using bars.

**Demonstrate the Concept**

The height of each bar represents the number of records in that category.

**Real-Life Example**

We can use a bar chart to compare the number of employees in each Department.

**AI/ML Usage**

Bar charts help us understand category distribution and identify categories with high or low frequencies.

```python
department_counts = df["Department"].value_counts()

# bar chart
plt.figure(figsize=(8, 5))
plt.bar(department_counts.index, department_counts.values)
plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

**Interpretation**

- The bar chart shows the number of employees in each department.
- Taller bars represent departments with more employees.

#### **21. Count Plot**



**Understand the Concept**

A count plot shows the number of records in each category.

**Demonstrate the Concept**

It automatically counts the values in a categorical column and displays them as bars.

**Real-Life Example**

We can use a count plot to see the number of employees in each Department.

**AI/ML Usage**

Count plots help us understand the distribution of categorical variables and identify categories with high or low frequencies.

```python
# count plot for Department
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Department")
plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

**Interpretation**

- The count plot shows the number of employees in each department.

#### **22. Pie Chart**



**Understand the Concept**

A pie chart shows the proportion of different categories as parts of a whole.

**Demonstrate the Concept**

Each slice represents the percentage or proportion of a category.

**Real-Life Example**

We can use a pie chart to show the percentage of employees in each Department.

**AI/ML Usage**

Pie charts help us understand the proportion of categorical data and check category distribution.

```python
department_counts = df["Department"].value_counts()

# pie chart
plt.figure(figsize=(7, 7))
plt.pie(department_counts.values, labels=department_counts.index,  autopct="%1.1f%%")
plt.title("Employee Distribution by Department")
plt.show()
```

**Interpretation**

- The pie chart shows the percentage of employees in each department.
