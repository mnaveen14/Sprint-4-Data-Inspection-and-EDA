## **1. Class Distribution**

**Understand the Concept**

Class distribution shows how the target variable is distributed across different classes.

**Demonstrate the Concept**

In the Loan Approval Dataset, the target variable is LoanStatus with two classes:

- Approved
- Rejected

Analyzing the class distribution helps understand how the target classes are represented in the dataset.

## **3. Class Percentages**

**Understand the Concept**

Class percentages show the proportion of observations belonging to each target class.

**Demonstrate the Concept**

**Real-Life Example**

Class percentages show what proportion of total loan applications were approved or rejected.

**AI/ML Usage**

Class percentages help identify target imbalance and determine whether techniques such as resampling or class weighting may be needed.

```python
# Calculate class percentages
class_percentages = (df["LoanStatus"].value_counts(normalize=True).mul(100).round(2))
print(class_percentages)
```

```python
# Visualize class percentages
class_percentages.plot(kind="bar", figsize=(7, 4))
plt.title("Class Percentage of Loan Status")
plt.xlabel("Loan Status")
plt.ylabel("Percentage")
plt.tight_layout()
plt.show()
```

**Interpretation**

- The output shows the percentage of Approved and Rejected loans.
- A large percentage difference may indicate class imbalance.

## **4. Class Imbalance**

**Understand the Concept**

Class imbalance occurs when one target class has significantly more observations than another class.

**Demonstrate the Concept**

For the Loan Approval Dataset, we compare the number of Approved and Rejected loans to determine whether the target variable is balanced or imbalanced.

**Real-Life Example**

If Approved loans greatly outnumber Rejected loans, a model may become biased toward predicting Approved.

**AI/ML Usage**

Class imbalance can affect model performance. Techniques such as class weighting, oversampling, or undersampling may be used when needed.

```python
# Check class imbalance

class_counts = df["LoanStatus"].value_counts()
majority_class = class_counts.idxmax()
minority_class = class_counts.idxmin()
imbalance_ratio = (class_counts.max() / class_counts.min())
print("Majority Class:", majority_class)
print("Minority Class:", minority_class)
print("Imbalance Ratio:", round(imbalance_ratio, 2))
```

```python
# Visualize class imbalance
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="LoanStatus")
plt.title("Class Imbalance in Loan Status")
plt.xlabel("Loan Status")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

**Interpretation**

- The majority class has more observations than the minority class.
- The imbalance ratio shows the difference between the two classes.
- A high imbalance ratio may require techniques such as class weighting or resampling.

## **2. Target Statistics and Outlier Analysis**

**Understand the Concept**

Statistical measures help us understand the center, spread, shape, and unusual values of the target variable.

For the House Price Dataset, we analyze:

- Mean
- Median
- Variance
- Standard Deviation
- Skewness
- Outliers

**Demonstrate the Concept**

These measures help us understand the overall behavior of HousePrice.

**AI/ML Usage**

Understanding these statistics helps identify skewness, high variability, and extreme values before training a regression model.

```python
# target statistics

target = df["HousePrice"]

print("Mean:", round(target.mean(), 2))
print("Median:", round(target.median(), 2))
print("Variance:", round(target.var(), 2))
print("Standard Deviation:", round(target.std(), 2))
print("Skewness:", round(target.skew(), 2))
```

```python
# Detect outliers using IQR

q1 = target.quantile(0.25)
q3 = target.quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = df[(df["HousePrice"] < lower_bound) | (df["HousePrice"] > upper_bound)]
print("Number of Outliers:", len(outliers))
```

```python
# Visualize outliers
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["HousePrice"])
plt.title("Outliers in House Price")
plt.xlabel("House Price")
plt.tight_layout()
plt.show()
```

**Interpretation**

- Mean and Median show the central value of HousePrice.
- Variance and Standard Deviation show how widely prices are spread.
- Skewness shows whether the target distribution is symmetric or skewed.
- The IQR method and box plot help identify unusual house prices.
